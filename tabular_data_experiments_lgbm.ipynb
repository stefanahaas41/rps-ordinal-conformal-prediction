```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

from util.tabular_lgbm_experiments_runner import run_experiment

datasets = {
    "heartDisease": "TOC-UCO/data/oc04_heartDisease.csv",
    "mammoexp": "TOC-UCO/data/oc03_mammoexp.csv",
    "support": "TOC-UCO/data/oc04_support.csv",
    "winequalityRed": "TOC-UCO/data/oc05_winequalityRed.csv",
    "nhanes": "TOC-UCO/data/oc05_nhanes.csv",
    "LEVXSensors": "TOC-UCO/data/oc04_LEVXSensors.csv",
}


for dataset, path in datasets.items():
    X = pd.read_csv(path, header=0)
    y = X.pop('y')
    le = LabelEncoder()
    le.fit(y)
    run_experiment(X, y, le, dataset)
```
